

# Part 1 – Statistical Mechanics of the Quantum Harmonic Oscillator

---

## Question 1. Partition Function

The energy eigenvalues of a quantum harmonic oscillator are

$
E_j=\hbar\omega\left(j+\frac12\right), \qquad j=0,1,2,\ldots
$

The canonical partition function is

$
Z(T)=\sum_{j=0}^{\infty}e^{-E_j/(k_BT)}.
$

Substituting the energy eigenvalues,

$
Z=\sum_{j=0}^{\infty}
e^{-\hbar\omega(j+\frac12)/(k_BT)}.
$

Factor out the constant term,

$
Z=e^{-\frac{\hbar\omega}{2k_BT}}
\sum_{j=0}^{\infty}
\left(e^{-\frac{\hbar\omega}{k_BT}}\right)^j.
$

Since

$
r=e^{-\frac{\hbar\omega}{k_BT}}<1,
$

the series is geometric:

$
\sum_{j=0}^{\infty}r^j=\frac1{1-r}.
$

Therefore,

$
\boxed{
Z(T)=
\frac{
e^{-\frac{\hbar\omega}{2k_BT}}
}
{
1-e^{-\frac{\hbar\omega}{k_BT}}
}
}
$

---

## Question 2(a): Helmholtz Free Energy

The Helmholtz free energy is defined as

\$
F=-k_B T\ln Z.
\$

Using the partition function,

$
\ln Z
=
-\frac{\hbar\omega}{2k_B T}
-
\ln\!\left(1-e^{-\frac{\hbar\omega}{k_B T}}\right).
$

Therefore,

$
F
=
-k_B T
\left[
-\frac{\hbar\omega}{2k_B T}
-
\ln\!\left(1-e^{-\frac{\hbar\omega}{k_B T}}\right)
\right].
$

Simplifying,

$
\boxed{
F
=
\frac{\hbar\omega}{2}
+
k_B T
\ln\!\left(1-e^{-\frac{\hbar\omega}{k_B T}}\right)
}
$



Hence,

$
\boxed{
F=
\frac{\hbar\omega}{2}
+
k_BT
\ln
\left(
1-
e^{-\frac{\hbar\omega}{k_BT}}
\right)
}
$

---

# Question 2(b): Internal Energy

The internal energy is defined as

$
U=-\frac{\partial \ln Z}{\partial \beta},
$

where

$
\beta=\frac{1}{k_B T}.
$

Since,

$
\ln Z
=
-\frac{\beta\hbar\omega}{2}
-
\ln\left(1-e^{-\beta\hbar\omega}\right),
$

differentiate with respect to \(\beta\):

$
\frac{\partial \ln Z}{\partial \beta}
=
-\frac{\hbar\omega}{2}
-
\frac{\hbar\omega e^{-\beta\hbar\omega}}
{1-e^{-\beta\hbar\omega}}.$

Thus,

$
U
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}
{e^{\beta\hbar\omega}-1}.
$

Replacing

$
\beta=\frac{1}{k_B T},
$

the internal energy becomes

$
\boxed{
U
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}
{e^{\frac{\hbar\omega}{k_B T}}-1}
}
$

---

# Question 3: Entropy

The entropy is defined as

$
S
=
-\left(\frac{\partial F}{\partial T}\right).
$

Using the expression for the Helmholtz free energy,

$
F
=
\frac{\hbar\omega}{2}
+
k_B T
\ln\left(1-e^{-\frac{\hbar\omega}{k_B T}}\right),
$

after differentiation, the entropy becomes

$
\boxed{
S
=
k_B
\left[
\frac{\frac{\hbar\omega}{k_B T}}
{e^{\frac{\hbar\omega}{k_B T}}-1}
-
\ln\left(1-e^{-\frac{\hbar\omega}{k_B T}}\right)
\right]
}
$

## Verification

Using the thermodynamic identity

\$
U=F+TS,
\$

and substituting the expressions for \(F\) and \(S\), we obtain

$
\boxed{
U
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}
{e^{\frac{\hbar\omega}{k_B T}}-1}
}
$


---

# Question 4: Heat Capacity

The heat capacity at constant volume is defined as

$
C_V
=
\left(\frac{\partial U}{\partial T}\right)_V.
$

Differentiating the internal energy expression gives

$
\boxed{
C_V
=
k_B
\left(
\frac{\hbar\omega}{k_B T}
\right)^2
\frac{
e^{\frac{\hbar\omega}{k_B T}}
}
{
\left(
e^{\frac{\hbar\omega}{k_B T}}-1
\right)^2
}
}
$

Using the identity

$
\sinh(x)
=
\frac{e^x-e^{-x}}{2},
$

the heat capacity can be simplified as

$
\boxed{
C_V
=
k_B
\left(
\frac{\hbar\omega}{2k_B T}
\right)^2
\frac{1}
{
\sinh^2
\left(
\frac{\hbar\omega}{2k_B T}
\right)
}
}
$

---

# Question 5. High-Temperature Limit

When

$
k_BT\gg\hbar\omega,
$

the exponential term can be approximated as

$
e^{\frac{\hbar\omega}{k_BT}}
\approx
1+\frac{\hbar\omega}{k_BT}.
$

Therefore,

$
C_V
\approx
k_B.
$

Hence,

$
\boxed{
\lim_{T\to\infty}C_V=k_B
}
$




# Part 2 – The Maximum Entropy Distribution under Mean and Variance Constraints

Given a continuous random variable \$X \in (-\infty,\infty)\$ with

\$
E[X]=\mu
\$

and

\$
E[(X-\mu)^2]=\sigma^2,
\$

show that the probability density that maximizes entropy is the Gaussian distribution.

---

# Step 1. Shannon Differential Entropy

The differential entropy of a continuous random variable is

$
H[p]
=
-\int_{-\infty}^{\infty} p(x)\ln p(x)\,dx.
$

Our objective is

$
\boxed{
\text{Maximize }
H[p]
=
-\int p(x)\ln p(x)\,dx
}
$

subject to the given constraints.

---

# Step 2. Constraints

## (a) Normalization

$
\int_{-\infty}^{\infty}p(x)\,dx=1.
$

---

## (b) Mean Constraint

$
\int_{-\infty}^{\infty}x p(x)\,dx=\mu.
$

---

## (c) Variance Constraint

$
\int_{-\infty}^{\infty}(x-\mu)^2p(x)\,dx=\sigma^2.
$

---

# Step 3. Construct the Lagrangian

Introduce Lagrange multipliers

$
\lambda_0,\lambda_1,\lambda_2.
$

The functional becomes

[
\begin{aligned}
\mathcal{L}
=&
-\int p(x)\ln p(x)\,dx
\\
&
-\lambda_0
\left(
\int p(x)\,dx-1
\right)
\\
&
-\lambda_1
\left(
\int xp(x)\,dx-\mu
\right)
\\
&
-\lambda_2
\left(
\int (x-\mu)^2p(x)\,dx-\sigma^2
\right).
\end{aligned}
\]

---

# Step 4. Differentiate

Take the functional derivative with respect to \(p(x)\):

$
\frac{\delta\mathcal L}{\delta p(x)}=0.
$

This gives

$
-\ln p(x)-1
-\lambda_0
-\lambda_1x
-\lambda_2(x-\mu)^2
=0.
$

Therefore,

$
\ln p(x)
=
-1
-\lambda_0
-\lambda_1x
-\lambda_2(x-\mu)^2.
$

Exponentiating both sides,

$
p(x)
=
e^{-1-\lambda_0}
e^{-\lambda_1x}
e^{-\lambda_2(x-\mu)^2}.
$

Let

$
A=e^{-1-\lambda_0},
$

then

$
\boxed{
p(x)=Ae^{-\lambda_1x-\lambda_2(x-\mu)^2}
}
$

---

# Step 5. Simplify the Expression

Since the mean is fixed at \(\mu\), the linear term is absorbed into the quadratic expression, giving

$
p(x)
=
Ae^{-\lambda(x-\mu)^2}.
$

Using the normalization condition,

$
\int_{-\infty}^{\infty}
Ae^{-\lambda(x-\mu)^2}
dx
=1.
$

Using the Gaussian integral,

$
\int_{-\infty}^{\infty}
e^{-\lambda(x-\mu)^2}dx
=
\sqrt{\frac{\pi}{\lambda}},
$

we obtain

$
A=\sqrt{\frac{\lambda}{\pi}}.
$

---

# Step 6. Determine \$\lambda\$

Apply the variance constraint,

$
\int_{-\infty}^{\infty}
(x-\mu)^2p(x)\,dx
=
\sigma^2.
$

For a Gaussian distribution,

$
\lambda=\frac{1}{2\sigma^2}.
$

Substituting into the normalization constant,

$
A
=
\frac{1}{\sqrt{2\pi\sigma^2}}
$

or

$
A
=
\frac{1}{\sqrt{2\pi}\sigma}.
$

---

# Final Maximum Entropy Distribution

Hence, the probability density function that maximizes entropy is

$
\boxed{
p(x)
=
\frac{1}{\sqrt{2\pi}\sigma}
\exp
\left[
-\frac{(x-\mu)^2}{2\sigma^2}
\right]
}
$

or equivalently,

$
\boxed{
X\sim N(\mu,\sigma^2)
}
$

---



# Part 3 – Multiclass Logistic Regression: Automatic Classification of Emails

## Problem Statement

An e-commerce platform automatically classifies incoming customer support emails into two classes:

- $(Y=0)$: Finance
- $(Y=1)$: Technical

The AI feature extractor produces the binary feature vector

$$
x=(x_1,x_2,x_3)^T
$$

where

- $x_1$: Financial keywords detected
- $x_2$: Account access keywords detected
- $x_3$: Bug or malfunction keywords detected

The observed feature probabilities are:

| Scenario | Feature Vector | Probability |
| -------- | -------------- | ----------- |
| A | $(1,0,0)$ | $0.40$ |
| B | $(0,1,0)$ | $0.25$ |
| C | $(0,0,1)$ | $0.05$ |
| D | $(0,1,1)$ | $0.15$ |
| E | $(1,1,0)$ | $0.05$ |

The historical constraints are:

$$
E[x_1I(Y=1)]=0.05
$$

$$
E[x_2I(Y=1)]=0.15
$$

$$
E[x_3I(Y=1)]=0.16
$$

---

# Step 1. Objective

Using the Maximum Entropy principle, determine the conditional probability

$$
P(Y=1|x)
$$

subject to the given constraints.

The conditional entropy is

$$
H(Y|X)
=
-\sum_xP(x)
\sum_y
P(y|x)
\ln P(y|x)
$$

The objective is

$$
\boxed{
\text{Maximize }H(Y|X)
}
$$

subject to the observed expectations.

---

# Step 2. Constraints

The feature expectation constraints are

$$
\sum_xP(x)x_1P(Y=1|x)=0.05
$$

$$
\sum_xP(x)x_2P(Y=1|x)=0.15
$$

$$
\sum_xP(x)x_3P(Y=1|x)=0.16
$$

Also,

$$
P(Y=0|x)+P(Y=1|x)=1
$$

---

# Step 3. Form the Lagrangian

Introduce Lagrange multipliers

$$
\lambda_1,\lambda_2,\lambda_3
$$

The optimization problem becomes

$$
\mathcal L
=
-H(Y|X)
+\lambda_1C_1
+\lambda_2C_2
+\lambda_3C_3
$$

where each $C_i$ represents one of the expectation constraints.

---

# Step 4. Differentiate

Differentiating the Lagrangian with respect to

$$
P(Y=1|x)
$$

gives

$$
\ln P(Y=1|x)
=
\lambda_1x_1
+\lambda_2x_2
+\lambda_3x_3
-\ln Z(x)
$$

where $Z(x)$ is the normalization constant.

Exponentiating both sides,

$$
P(Y=1|x)
=
\frac{
e^{\lambda_1x_1+\lambda_2x_2+\lambda_3x_3}
}
{Z(x)}
$$

---

# Step 5. Compute the Partition Function

Since there are only two classes,

$$
Y=0
$$

and

$$
Y=1
$$

the partition function is

$$
Z(x)
=
1+
e^{\lambda_1x_1+\lambda_2x_2+\lambda_3x_3}
$$

---

# Final Softmax (Logistic) Model

Therefore,

$$
\boxed{
P(Y=1|x)
=
\frac{
e^{\lambda_1x_1+\lambda_2x_2+\lambda_3x_3}
}
{
1+
e^{\lambda_1x_1+\lambda_2x_2+\lambda_3x_3}
}
}
$$

or equivalently,

$$
\boxed{
P(Y=1|x)
=
\frac{1}
{
1+
e^{-(\lambda_1x_1+\lambda_2x_2+\lambda_3x_3)}
}
}
$$

This is the **logistic regression (sigmoid)** model.

Similarly,

$$
\boxed{
P(Y=0|x)=1-P(Y=1|x)
}
$$

---

# Interpretation of the Constraints

## Constraint 1

$$
E[x_1I(Y=1)]=0.05
$$

Financial keywords rarely lead to Technical Support. Therefore,

$$
\boxed{\lambda_1<0}
$$

A negative weight decreases the probability of assigning an email to Technical Support.

---

## Constraint 2

$$
E[x_2I(Y=1)]=0.15
$$

Account access issues are sometimes routed to Technical Support and sometimes to Finance.

Therefore,

$$
\boxed{\lambda_2>0}
$$

but its value is moderate.

---

## Constraint 3

$$
E[x_3I(Y=1)]=0.16
$$

Bug-related keywords strongly indicate Technical Support.

Therefore,

$$
\boxed{\lambda_3>0}
$$

and it is expected to be the largest positive coefficient.

---

# Interpretation of the Scenarios

## Scenario A

$$
x=(1,0,0)
$$

Contains only financial keywords.

Since

$$
\lambda_1<0
$$

$$
\boxed{
P(Y=1|x)\text{ is very small}
}
$$

The email is most likely classified as **Finance**.

---

## Scenario B

$$
x=(0,1,0)
$$

Contains account access keywords.

$$
\boxed{
P(Y=1|x)\text{ is moderate}
}
$$

The email could be routed to either department depending on the learned parameters.

---

## Scenario C

$$
x=(0,0,1)
$$

Contains bug-related keywords.

$$
\boxed{
P(Y=1|x)\text{ is high}
}
$$

The email is most likely classified as **Technical Support**.

---

## Scenario D

$$
x=(0,1,1)
$$

Contains both account access and bug-related keywords.

Since both coefficients are positive,

$$
\boxed{
P(Y=1|x)\text{ is very high}
}
$$

This email is strongly classified as **Technical Support**.

---

## Scenario E

$$
x=(1,1,0)
$$

Contains financial and account access keywords.

The positive and negative coefficients compete with each other, resulting in

$$
\boxed{
P(Y=1|x)\text{ is moderate or relatively low}
}
$$

The final classification depends on the learned parameter values.

---



# Part 4 – Asset Pricing and Portfolio Optimization using the Maximum Entropy Principle

## Background

Suppose the future asset price $S_T>0$ is unknown. Instead of assuming a Gaussian distribution, we infer the probability density using the **Maximum Entropy (MaxEnt)** principle subject only to the known market constraints.

---

# Step 1. Objective Function

For a continuous random variable $S_T$, the differential entropy is

$$
H[p]
=
-\int_{0}^{\infty}p(S_T)\ln p(S_T)\,dS_T.
$$

The objective is

$$
\boxed{
\text{Maximize }H[p]
}
$$

subject to the observed market information.

---

# Step 2. Constraints

## (a) Normalization

$$
\int_{0}^{\infty}p(S_T)\,dS_T=1.
$$

---

## (b) Expected Future Price

The expected future asset price equals the forward price,

$$
\boxed{
E[S_T]=102
}
$$

or

$$
\int_{0}^{\infty}
S_Tp(S_T)\,dS_T
=
102.
$$

---

## (c) Expected Squared Price

The second moment is constrained by

$$
\boxed{
E[S_T^2]=10600
}
$$

or

$$
\int_{0}^{\infty}
S_T^2p(S_T)\,dS_T
=
10600.
$$

---

## (d) Logarithmic (Crash-Risk) Constraint

The market also specifies

$$
\boxed{
E[\ln(S_T)]=4.60
}
$$

or

$$
\int_{0}^{\infty}
\ln(S_T)p(S_T)\,dS_T
=
4.60.
$$

---

# Step 3. Construct the Lagrangian

Introduce Lagrange multipliers

$$
\lambda_0,\lambda_1,\lambda_2,\lambda_3.
$$

The Lagrangian is

$$
\begin{aligned}
\mathcal L
=&
-\int p(S_T)\ln p(S_T)\,dS_T
\\
&
-\lambda_0
\left(
\int p(S_T)\,dS_T-1
\right)
\\
&
-\lambda_1
\left(
\int S_Tp(S_T)\,dS_T-102
\right)
\\
&
-\lambda_2
\left(
\int S_T^2p(S_T)\,dS_T-10600
\right)
\\
&
-\lambda_3
\left(
\int \ln(S_T)p(S_T)\,dS_T-4.60
\right).
\end{aligned}
$$

---

# Step 4. Differentiate

Taking the functional derivative,

$$
\frac{\delta\mathcal L}
{\delta p(S_T)}
=0,
$$

gives

$$
-\ln p(S_T)-1
-\lambda_0
-\lambda_1S_T
-\lambda_2S_T^2
-\lambda_3\ln(S_T)
=0.
$$

Therefore,

$$
\ln p(S_T)
=
-1
-\lambda_0
-\lambda_1S_T
-\lambda_2S_T^2
-\lambda_3\ln(S_T).
$$

Exponentiating,

$$
p(S_T)
=
e^{-1-\lambda_0}
S_T^{-\lambda_3}
e^{-\lambda_1S_T-\lambda_2S_T^2}.
$$

---

# Step 5. Partition Function

Let

$$
Z=e^{1+\lambda_0}.
$$

Then

$$
\boxed{
p(S_T)
=
\frac{1}{Z}
S_T^{-\lambda_3}
e^{-\lambda_1S_T-\lambda_2S_T^2}
}
$$

where

$$
\boxed{
Z
=
\int_{0}^{\infty}
S_T^{-\lambda_3}
e^{-\lambda_1S_T-\lambda_2S_T^2}
\,dS_T
}
$$

This normalization constant ensures that

$$
\int_0^\infty p(S_T)\,dS_T=1.
$$

---

# Final Maximum Entropy Probability Density

Thus, the least-biased probability density satisfying all market constraints is

$$
\boxed{
p(S_T)
=
\frac{
S_T^{-\lambda_3}
e^{-\lambda_1S_T-\lambda_2S_T^2}
}
{
\displaystyle
\int_0^\infty
S_T^{-\lambda_3}
e^{-\lambda_1S_T-\lambda_2S_T^2}
\,dS_T
}
}
$$
